```python
# Matrix Multiplication has strict
# dimension rules.
tensor_A = torch.tensor([[1, 2],
                        [3, 4],
                        [5, 6]], dtype=torch.float32)
tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]], dtype=torch.float32)
# torch.matmul(tensor_A, tensor_B) # This should fail
```

```python
# View tensor A and B
print(tensor_A)
print(tensor_B)
```

```python
print(tensor_A)
print(tensor_B.T)
```

```python
# If we transpose, the operation works
print(f"Original tensor shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shapes: tensor_A = {tensor_A.shape}, tensor_B.T = {tensor_B.T.shape}")
print(f"Multiplying: {tensor_A.shape} * {tensor_B.T.shape} ")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"Output shape: {output.shape}")
```

```python
# torch.mm is short for matmul
torch.mm(tensor_A, tensor_B.T)
```

```python
# Linear layers starts with a random weights matrix,
# so lets make it reproducible:
torch.manual_seed(42)
linear = torch.nn.Linear(in_features=2, # matches inner dimension of input
                         out_features=6) # describes outer valu
x = tensor_A
output = linear(x)
print(f"Input shape: {x.shape}")
print(f"Output: {output}")
print(f"Output shape: {output.shape}")

# Tweaking it a bit...
linear = torch.nn.Linear(in_features=3,
                         out_features=6)
x = tensor_A.T
output = linear(x)
print(f"\nInput shape: {x.shape}")
print(f"Output: {output}")
print(f"Output shape: {output.shape}")
```

```python
x = torch.arange(1., 8.)
x, x.shape
```

```python
x_reshaped = x.reshape((1, 7))
x_reshaped, x_reshaped.shape
```

```python
z = x.view(1, 7)
z, z.shape
```

```python
# Changing z changes x, as z is just a different view
z[:,0] = 5
z, x
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x], dim=0)
x_stacked
```

```python
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape : {x_reshaped.shape}")

# Remove extra dimension from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
print(f"Previous tensor: {x_squeezed}")
print(f"Previous tensor shape: {x_squeezed.shape}")

x_unsqueezed = x_squeezed.unsqueeze(0)
print(f"New vector: {x_unsqueezed}")
print(f"New vector shape: {x_unsqueezed.shape}")
```

```python
# New vector
x_original = torch.rand(size=(244, 244, 3))

x_permuted = x_original.permute(2, 0, 1) # x_permuted is NOT a new vector, its a view.
print(f"Previous shape: {x_original.shape}")
print(f"New shape     : {x_permuted.shape}")
```

```python
# Two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)
print(f"random_tensor_A: {random_tensor_A}")
print(f"random_tensor_B: {random_tensor_B}")
print(f"Does A equal B anywhere?")
random_tensor_A == random_tensor_B
```

```python
# Set random seed
RANDOM_SEED = 69420
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.random.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(f"random_tensor_C : {random_tensor_C}")
print(f"random_tensor_D: {random_tensor_D}")
print("Are C and D equal anywhere?")
random_tensor_C == random_tensor_D
```
